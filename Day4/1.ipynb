```python
df = df.drop(columns=['Make', 'Model', 'Vehicle Class'])
df
```

```python
df= df.drop(['Fuel Consumption City (L/100 km)','Fuel Consumption Hwy (L/100 km)','Fuel Consumption Comb (L/100 km)','Fuel Consumption Comb (mpg)'], axis=1)
df
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X = df[["Engine Size(L)", "Cylinders"]]
Y = df["CO2 Emissions(g/km)"]

Y
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split( X, Y, test_size=0.2, random_state=42)
```

```python
(x_train.shape)
(x_test.shape)
(y_train.shape)
(y_test.shape)
```

```python
model = LinearRegression()
```

```python
model.fit(x_train, y_train)
```

```python
y_predict= model.predict(x_test)
y_predict
```

```python
r2_score(y_test,y_predict)
```
