```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)

age = np.random.randint(25, 56, 100)

print(age)
```

```python
salary_type = np.random.choice(
    ['Lower', 'Medium', 'High'], 100
)

print(salary_type)
```

```python
salary = []

for category in salary_type:

    if category == 'Lower':
        salary.append(np.random.randint(15000, 30001))

    elif category == 'Medium':
        salary.append(np.random.randint(30000, 60001))

    else:
        salary.append(np.random.randint(60000, 120001))

print(salary)
```

```python
df = pd.DataFrame({
    'Age': age,
    'Salary': salary,
    'Salary_Type': salary_type
})

print(df)
```

```python
df.head()
```

```python
plt.scatter(df['Age'], df['Salary'])

plt.xlabel('Age')
plt.ylabel('Salary')
plt.title('Age vs Salary')

plt.show()
```

```python
df.to_csv('salary_dataset.csv', index=False)
```

```python
from sklearn.cluster import KMeans
```

```python
X = df[['Age', 'Salary']]
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

kmeans.fit(X)
```

```python
df['Cluster'] = kmeans.labels_

print(df.head(10))
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df['Age'],
    df['Salary'],
    c=df['Cluster'],
    cmap='viridis'
)

plt.xlabel('Age')
plt.ylabel('Salary')
plt.title('K-Means Clustering: Age vs Salary')

plt.show()
```

```python
print("Cluster Centers:")
print(kmeans.cluster_centers_)
```
