# Day 28: Lasso (L1) Regularization & Feature Sparsity
Loss = MSE + alpha * sum(|w_i|)

```python
from sklearn.linear_model import Lasso
from sklearn.datasets import make_regression
import numpy as np

X, y = make_regression(n_samples=50, n_features=10, n_informative=3, noise=1.0, random_state=42)
lasso = Lasso(alpha=0.5)
lasso.fit(X, y)

print('Zeroed coefficients count:', np.sum(lasso.coef_ == 0))
print('Coefficients:', np.round(lasso.coef_, 2))
```
