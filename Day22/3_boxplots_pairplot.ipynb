# Day 22: Boxplots and Pairplots for EDA

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Category': np.random.choice(['Group 1', 'Group 2', 'Group 3'], 150),
    'Metric_X': np.random.normal(50, 15, 150),
    'Metric_Y': np.random.normal(100, 25, 150)
})

plt.figure(figsize=(8, 4))
sns.boxplot(x='Category', y='Metric_X', data=df, palette='Set2')
plt.title('Distribution and Outliers across Categories')
plt.show()
```
