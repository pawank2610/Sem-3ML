# Day 22: Feature Correlation Heatmaps with Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

np.random.seed(42)
features = pd.DataFrame(np.random.randn(100, 4), columns=['Feature_A', 'Feature_B', 'Feature_C', 'Feature_D'])
corr = features.corr()

plt.figure(figsize=(7, 5))
sns.heatmap(corr, annot=True, cmap='coolwarm', vmin=-1, vmax=1, fmt='.2f')
plt.title('Feature Correlation Matrix')
plt.show()
```
