# Day 22: Statistical Visualizations with Seaborn
Distribution plots, KDE, and count plots.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

sns.set_theme(style='whitegrid')
tips = sns.load_dataset('tips') if 'tips' in sns.get_dataset_names() else pd.DataFrame({
    'total_bill': np.random.uniform(10, 50, 100),
    'tip': np.random.uniform(2, 10, 100),
    'day': np.random.choice(['Thur', 'Fri', 'Sat', 'Sun'], 100)
})

plt.figure(figsize=(8, 4))
sns.histplot(data=tips, x='total_bill', kde=True, color='indigo')
plt.title('Bill Amount Distribution with KDE')
plt.show()
```
