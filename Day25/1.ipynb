# Day 25: Feature Scaling with StandardScaler and MinMaxScaler

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler

data = pd.DataFrame({
    'Salary': [30000, 45000, 70000, 120000, 55000],
    'Age': [22, 28, 35, 50, 31]
})

# StandardScaler (Mean = 0, Variance = 1)
std_scaler = StandardScaler()
scaled_std = std_scaler.fit_transform(data)

# MinMaxScaler (Range [0, 1])
minmax_scaler = MinMaxScaler()
scaled_minmax = minmax_scaler.fit_transform(data)

print('Standard Scaled:\n', scaled_std)
print('\nMinMax Scaled:\n', scaled_minmax)
```
