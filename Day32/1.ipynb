# Day 32: Support Vector Classifier (SVC) Experiments

```python
from sklearn.svm import SVC
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split

X, y = make_moons(n_samples=200, noise=0.2, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

svm_linear = SVC(kernel='linear').fit(X_train, y_train)
svm_rbf = SVC(kernel='rbf', C=1.0, gamma='scale').fit(X_train, y_train)

print('Linear Kernel Accuracy:', svm_linear.score(X_test, y_test))
print('RBF Kernel Accuracy:', svm_rbf.score(X_test, y_test))
```
