# Day 23: Data Distribution, Skewness & Transformations

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Right skewed variable
right_skewed = np.random.exponential(scale=10, size=500)
log_transformed = np.log1p(right_skewed)

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].hist(right_skewed, bins=25, color='salmon', edgecolor='black')
axes[0].set_title(f'Original (Skew: {pd.Series(right_skewed).skew():.2f})')
axes[1].hist(log_transformed, bins=25, color='mediumseagreen', edgecolor='black')
axes[1].set_title(f'Log Transformed (Skew: {pd.Series(log_transformed).skew():.2f})')
plt.tight_layout()
plt.show()
```
