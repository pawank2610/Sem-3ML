# Day 23: Exploratory Data Analysis (EDA) Workflow
Standard routine to explore datasets before modeling.

```python
import pandas as pd
import numpy as np

# Synthetic Customer Dataset
np.random.seed(101)
n = 200
df = pd.DataFrame({
    'Age': np.random.randint(18, 70, n),
    'Annual_Income_k': np.random.uniform(20, 120, n),
    'Spending_Score': np.random.randint(1, 100, n),
    'Membership': np.random.choice(['Bronze', 'Silver', 'Gold'], n)
})
print('Dataset Shape:', df.shape)
print('\nInfo:')
print(df.info())
print('\nSummary Statistics:')
print(df.describe())
```
