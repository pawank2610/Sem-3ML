# Day 19: NumPy Fundamentals and Array Operations
Creating multi-dimensional arrays, indexing, and slicing.

```python
import numpy as np

# Array initializations
a = np.array([1, 2, 3, 4, 5])
b = np.zeros((3, 3))
c = np.ones((2, 4))
d = np.arange(1, 13).reshape(3, 4)

print('1D Array:', a)
print('Zeros 3x3:\n', b)
print('Reshaped 3x4:\n', d)
```

```python
# Slicing & Indexing
print('Submatrix (first 2 rows, cols 1-3):\n', d[:2, 1:3])
print('Elements greater than 5:', d[d > 5])
```
