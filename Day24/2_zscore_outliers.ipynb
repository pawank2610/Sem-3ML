# Day 24: Outlier Detection using Z-Score
Points with |z| > 3 are typically flagged as outliers.

```python
import numpy as np
import pandas as pd
from scipy import stats

data = np.array([22, 24, 25, 23, 22, 26, 24, 23, 85, 24, 25, -15, 23])
mean = np.mean(data)
std = np.std(data)
z_scores = (data - mean) / std

outlier_indices = np.where(np.abs(z_scores) > 2.5)[0]
print('Outlier values:', data[outlier_indices])
```
