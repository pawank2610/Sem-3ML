# Day 24: Outlier Detection using the IQR Method
Formula: IQR = Q3 - Q1, Bounds = [Q1 - 1.5*IQR, Q3 + 1.5*IQR]

```python
import numpy as np
import pandas as pd

np.random.seed(42)
data = np.concatenate([np.random.normal(50, 5, 100), [10, 95, 100, 105]])
s = pd.Series(data)

q1 = s.quantile(0.25)
q3 = s.quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = s[(s < lower_bound) | (s > upper_bound)]
print('Lower bound:', lower_bound)
print('Upper bound:', upper_bound)
print('Detected outliers:\n', outliers.values)
```
