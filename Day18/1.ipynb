# Day 18: Advanced Data Cleaning with Pandas
Handling missing data, imputation strategies, and interpolations.

```python
import pandas as pd
import numpy as np

# Create synthetic dataset with missing values
data = {
    'Student_ID': [101, 102, 103, 104, 105, 106],
    'Study_Hours': [4.5, np.nan, 3.2, 5.8, np.nan, 6.1],
    'Attendance_Rate': [90, 85, np.nan, 95, 78, 88],
    'Exam_Score': [78, 82, 65, np.nan, 70, 92]
}
df = pd.DataFrame(data)
df
```

```python
# Identify missing values
print('Null count per column:')
print(df.isnull().sum())

# Mean and Median Imputation
df_imputed = df.copy()
df_imputed['Study_Hours'] = df_imputed['Study_Hours'].fillna(df_imputed['Study_Hours'].median())
df_imputed['Attendance_Rate'] = df_imputed['Attendance_Rate'].fillna(df_imputed['Attendance_Rate'].mean())
df_imputed['Exam_Score'] = df_imputed['Exam_Score'].interpolate(method='linear')
df_imputed
```
