```python
import numpy as np 
import pandas as pd
```

```python
data= np.array([1,2,3,4,7,89,0,87,4,2,3,5,6,6,7,8,3,2,2,4,677,955,765])
q1=np.percentile(data,25)
q2=np.percentile(data,50)
q3=np.percentile(data,75)

print(q1)
print(q2)
print(q3)
```

```python
df=pd.DataFrame({"Values":data})
print(df)
```

```python
df.info
df.describe()
```

```python
Q1=df.quantile(0.25)
Q2=df.quantile(0.50)
Q3=df.quantile(0.75)

Q3
```

```python
IQR=Q3-Q1
IQR
```

```python
lower=Q1-IQR*0.25
upper=Q3+IQR*0.75
print(upper)
print(lower)
```

```python
# outlier
lower_val = lower["Values"]  # or Lower.item()
upper_val = upper["Values"]  # or Upper.item()

outliers = df[
    (df["Values"] < lower_val) | 
    (df["Values"] > upper_val)
]["Values"]

print(outliers.tolist())
```
