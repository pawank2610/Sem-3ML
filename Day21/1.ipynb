# Day 21: Data Visualization with Matplotlib
Line plots, scatter plots, and custom styling.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y1, label='sin(x)', color='royalblue', lw=2)
plt.plot(x, y2, label='cos(x)', color='darkorange', linestyle='--', lw=2)
plt.title('Trigonometric Waveforms')
plt.xlabel('X (radians)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
