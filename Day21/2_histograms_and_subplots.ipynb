# Day 21: Matplotlib Subplots and Histograms
Using Figure and Axes object-oriented interface.

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
data_normal = np.random.normal(loc=50, scale=10, size=1000)
data_exp = np.random.exponential(scale=2, size=1000)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(data_normal, bins=30, color='teal', edgecolor='black', alpha=0.7)
axes[0].set_title('Gaussian Distribution')
axes[1].hist(data_exp, bins=30, color='coral', edgecolor='black', alpha=0.7)
axes[1].set_title('Exponential Distribution')
plt.tight_layout()
plt.show()
```
